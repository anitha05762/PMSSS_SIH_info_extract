```python
from pdf2image import convert_from_path
import pytesseract
import os

def pdf_to_images(pdf_path, output_folder, dpi=300, output_format='png'):
    """
    Convert a PDF file to images.

    :param pdf_path: Path to the PDF file.
    :param output_folder: Folder where the images will be saved.
    :param dpi: DPI (dots per inch) for the output images. Default is 300.
    :param output_format: Format of the output images. Default is 'png'.
    :return: List of paths to the saved images.
    """
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)  # Create the folder if it doesn't exist

    images = convert_from_path(pdf_path, dpi=dpi)  # Convert PDF to images
    image_paths = []

    for i, image in enumerate(images):
        image_path = os.path.join(output_folder, f'page_{i + 1}.{output_format}')
        image.save(image_path, output_format.upper())  # Save each image
        image_paths.append(image_path)

    return image_paths

def extract_text_from_image(image_path):
    """
    Extract text from an image using Tesseract OCR.

    :param image_path: Path to the image.
    :return: Extracted text.
    """
    text = pytesseract.image_to_string(image_path)  # Extract text from image
    return text

def pdf_to_text(pdf_path, output_folder):
    """
    Convert a PDF to images and extract text from those images.

    :param pdf_path: Path to the PDF file.
    :param output_folder: Folder where the images and text will be saved.
    :return: None
    """
    image_paths = pdf_to_images(pdf_path, output_folder)  # Convert PDF to images
    all_text = ""

    for image_path in image_paths:
        text = extract_text_from_image(image_path)  # Extract text from each image
        all_text += text + "\n"

    text_output_path = os.path.join(output_folder, "extracted_text.txt")
    with open(text_output_path, "w") as text_file:
        text_file.write(all_text)  # Save extracted text to file

    print(f"Text extracted and saved at: {text_output_path}")

# Example usage:
pdf_path = '/content/sample_data/Ani_Community_certificate.pdf'
output_folder = 'output_images'
pdf_to_text(pdf_path, output_folder)
```

```python
# code 1
import pdfplumber
import re

# Open the PDF file
pdf_path = '/content/sample_data/12 marksheet anitha.pdf'  # Replace with your PDF file path

# Initialize variables
total_marks = None

with pdfplumber.open(pdf_path) as pdf:
    # Iterate through all pages
    for page_num, page in enumerate(pdf.pages):
        text = page.extract_text()

        # Check if "TOTAL MARKS" is in the text
        if "TOTAL MARKS" in text:
            # Find the line containing "TOTAL MARKS"
            lines = text.split('\n')
            for line in lines:
                if "TOTAL MARKS" in line:
                    # Extract total marks after colon ":"
                    total_marks = line.split(":")[-1].strip()
                    break

# Compute percentage if total marks are found
if total_marks:
    # Extract only the digits using regex
    numeric_marks = re.findall(r'\d+', total_marks)
    if numeric_marks:
        total_marks_int = int(''.join(numeric_marks))  # Join extracted digits and convert to integer
        max_marks = 600  # Assuming the maximum marks are 600
        percentage = (total_marks_int / max_marks) * 100

        print(f"Total Marks: {total_marks_int}")
        print(f"Percentage: {percentage:.2f}%")
    else:
        print("No numeric marks found in the total marks string.")
else:
    print("Total marks not found in the PDF.")
if (percentage >= 90):
  credit = 10
elif (percentage >= 80):
  credit = 9
elif (percentage >= 70):
  credit = 8
elif (percentage >= 60):
  credit = 7
elif (percentage >= 50):
  credit = 6
else:
  credit = 3
print ("credit:",credit)
```

```python
# code 3
import fitz  # PyMuPDF
import pytesseract
from PIL import Image
import re
import io

def extract_income_from_pdf(pdf_path):
    # Open the PDF
    doc = fitz.open(pdf_path)

    # Loop through all the pages
    for page_num in range(len(doc)):
        page = doc.load_page(page_num)

        # Extract images from the page
        images = page.get_images(full=True)

        for img_index, img in enumerate(images):
            # Get the XREF of the image
            xref = img[0]

            # Extract the image bytes
            base_image = doc.extract_image(xref)
            image_bytes = base_image["image"]
            image_ext = base_image["ext"]

            # Open the image with PIL
            image = Image.open(io.BytesIO(image_bytes))

            # Use pytesseract to extract text from the image
            text = pytesseract.image_to_string(image)

            # Print the extracted text to inspect
            #print(f"Extracted Text from image {img_index+1} on page {page_num+1}:\n", text)

            # Use a regular expression to find the specific text containing "Rs."
            pattern = r'Rs\.\s?\d+,?\d*'
            match = re.search(pattern, text)

            if match:
                print(f"Extracted income text on page {page_num+1}: {match.group()}")
                return match.group()  # Return the first match found

    print("No income text found.")
    return None

# Replace 'path_to_pdf' with your actual PDF file path
pdf_path = '/content/sample_data/sam income certficate.pdf'
income_text = extract_income_from_pdf(pdf_path)

if income_text:
    print(f"Extracted Income: {income_text}")
else:
    print("Income could not be extracted.")
```
